# Projet MLA : Analyse des offres d'emploi IT au Maroc
## Notebook 4 : Pipeline d'Inférence

Ce notebook montre comment utiliser le modèle entraîné pour prédire le niveau requis (Junior, Confirmé, Senior) d'une nouvelle offre d'emploi IT à partir de son titre et de sa description.

### 1. Chargement du prédicteur

```
import os
import sys
from src.models.predict import JobLevelPredictor

try:
    predictor = JobLevelPredictor()
    print("Modèle de prédiction chargé avec succès !")
except Exception as e:
    print(f"Erreur de chargement (le modèle doit être d'abord entraîné via train.py ou le Notebook 3) : {e}")
```

### 2. Test d'inférence interactif
Définissez un titre de poste et une description pour évaluer le niveau prédit.

```
test_title = "Développeur Fullstack React / Node.js Senior"
test_desc = """
Nous recherchons un expert Fullstack pour piloter l'architecture de nos projets.
Vous possédez au moins 6 ans d'expérience sur React, Node.js et AWS.
Capacité à encadrer une équipe technique (lead dev), concevoir des solutions robustes et animer les rituels agiles.
"""

res = predictor.predict(test_title, test_desc)
print(f"Niveau prédit : {res['predicted_level']}")
if res['probabilities']:
    print("Probabilités :")
    for label, prob in res['probabilities'].items():
        print(f"  - {label} : {prob*100:.2f}%")
```

### 3. Exemples comparatifs

```
exemples = [
    {
        "title": "Ingénieur DevOps Débutant",
        "desc": "Rattaché à l'équipe Infra, vous intervenez sur Docker, Kubernetes et Git. Jeune diplômé accepté avec stage de fin d'études significatif."
    },
    {
        "title": "Lead QA Testeur",
        "desc": "Nous cherchons un profil confirmé (4 ans d'expérience) pour rédiger les plans de test, implémenter Selenium et assurer la qualité logicielle."
    }
]

for idx, ex in enumerate(exemples):
    r = predictor.predict(ex['title'], ex['desc'])
    print(f"Exemple {idx+1} : {ex['title']}")
    print(f"  Prediction : {r['predicted_level']}")
    if r['probabilities']:
        probs_str = ", ".join([f"{k}: {v*100:.1f}%" for k, v in r['probabilities'].items()])
        print(f"  Confidence : {probs_str}")
    print()
```
