### 1. Importation des librairies et chargement des données

```
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from src.preprocessing.clean_data import DataCleaner

sns.set_theme(style="whitegrid")
df_raw_path = "data/raw/jobs_raw_latest.csv"
if os.path.exists(df_raw_path):
    df_raw = pd.read_csv(df_raw_path)
    print(f"Données brutes chargées : {len(df_raw)} lignes")
else:
    print("Fichier de données brutes introuvable. Exécutez d'abord le scraping.")
```

### 2. Nettoyage et Traitement NLP
Nous appliquons la classe `DataCleaner` pour filtrer les doublons, éliminer le bruit textuel, lemmatiser les mots, et générer le niveau de poste cible (`Niveau_du_poste`).

```
cleaner = DataCleaner()
df_cleaned = cleaner.process()
if df_cleaned is not None:
    display(df_cleaned[['titre_poste', 'Niveau_du_poste', 'nombre_technologies']].head(5))
```

### 3. Visualisations et Statistiques Descriptives
Analysons maintenant les distributions des postes, des villes, et des technologies les plus demandées au Maroc.

#### A. Distribution de la variable cible `Niveau_du_poste`

```
plt.figure(figsize=(7, 4))
sns.countplot(data=df_cleaned, x='Niveau_du_poste', palette='viridis', order=['Junior', 'Confirmé', 'Senior'])
plt.title("Distribution du Niveau des Postes IT au Maroc")
plt.xlabel("Niveau")
plt.ylabel("Nombre d'offres")
plt.show()
```

#### B. Répartition géographique des offres (Top Villes)

```
top_villes = df_cleaned['ville'].value_counts().head(10)
plt.figure(figsize=(10, 5))
sns.barplot(x=top_villes.values, y=top_villes.index, palette='mako')
plt.title("Top 10 des Villes avec le plus d'Offres d'Emploi IT")
plt.xlabel("Nombre d'offres")
plt.ylabel("Ville")
plt.show()
```

#### C. Technologies les plus demandées

```
tech_cols = [col for col in df_cleaned.columns if col.startswith('tech_')]
tech_counts = df_cleaned[tech_cols].sum().sort_values(ascending=False).head(15)
tech_names = [c.replace('tech_', '').upper() for c in tech_counts.index]

plt.figure(figsize=(10, 5))
sns.barplot(x=tech_counts.values, y=tech_names, palette='rocket')
plt.title("Top 15 des Technologies les Plus Demandées")
plt.xlabel("Nombre d'offres")
plt.ylabel("Technologie")
plt.show()
```

#### D. Types de Contrats Proposés

```
contrats = df_cleaned['type_contrat'].value_counts().head(5)
plt.figure(figsize=(6, 6))
plt.pie(contrats.values, labels=contrats.index, autopct='%1.1f%%', colors=sns.color_palette('pastel'), startangle=140)
plt.title("Répartition des Types de Contrats")
plt.show()
```
